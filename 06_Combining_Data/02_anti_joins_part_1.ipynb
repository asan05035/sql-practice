# 01_Left_anti_join

## Q1: Find all customers without orders

```sql
SELECT 
    c.CustomerID,
    c.FirstName,
    o.OrderID,
    o.ProductID
FROM workspace.salesdb.sales_customers AS c
LEFT JOIN workspace.salesdb.sales_orders AS o 
ON c.CustomerID = o.CustomerID
WHERE o.CustomerID IS NULL
```

# 02_Full_anti_join

## Q2: Find all customers without orders and all orders without customers

```sql
SELECT 
    c.CustomerID,
    c.FirstName,
    o.OrderID,
    o.ProductID
FROM workspace.salesdb.sales_customers AS c
FULL JOIN workspace.salesdb.sales_orders AS o 
ON c.CustomerID = o.CustomerID
WHERE o.CustomerID IS NULL OR c.CustomerID IS NULL
```

```sql
SELECT 
    c.CustomerID,
    c.FirstName,
    o.OrderID,
    o.ProductID
FROM workspace.salesdb.sales_customers AS c
JOIN workspace.salesdb.sales_orders AS o 
ON c.CustomerID = o.CustomerID
WHERE o.CustomerID IS NOT NULL
```

# 03_WITHOUT USING INNER JOIN

## Q3: Fill customers along with orders but only customers who placed an order

```sql
SELECT 
    c.CustomerID,
    c.FirstName,
    o.OrderID,
    o.ProductID
FROM workspace.salesdb.sales_customers AS c
INNER JOIN workspace.salesdb.sales_orders AS o
ON c.CustomerID = o.CustomerID
```

# 04_CROSS JOIN( *CARTESIAN JOIN* )

## Every possible combination i.e every row from left table with every row from right

```sql
SELECT
    c.CustomerID,
    c.FirstName,
    o.OrderID,
    o.ProductID
FROM workspace.salesdb.sales_customers AS c
CROSS JOIN workspace.salesdb.sales_orders AS o
```
